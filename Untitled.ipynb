```python
from pathlib import Path
import json

import pandas as pd

path = Path("data/user_data_tiktok.json")

with open(path, encoding="utf-8") as f:
    data = json.load(f)

print("done")
```

```python
activity = data["Your Activity"]

watchHistoryList = data["Your Activity"]["Watch History"]["VideoList"]

df = pd.DataFrame(watchHistoryList)          
```

```python
#df.head() -> returns all the headings
#df.shape -> returns # of col and rows
#df.dtypes -> returns the types of each field
df.info # Kind of shows everything
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df.dtypes

#df['Date'].head() this will show the first 5-rows, useful to know.
```

```python
df["Hour"] = df["Date"].dt.hour
df["DateOnly"] = df["Date"].dt.date
df["Day_Name"] = df["Date"].dt.day_name()

print(df.head())
```

```python
df['Hour'].value_counts()
```

```python
#df['DateOnly'].value_counts()

dailyCount = df['DateOnly'].value_counts().sort_index()
print(dailyCount)

 
weekly_counts = df.resample("W", on="Date").size() # 'W" is weekly frequency which is casted on date. Size function will then represent how many entries are
#in a week

#.plot creates a plot (Pandas method but uses Matplotlib)
ax = weekly_counts.plot(
    kind="bar",
    figsize=(12, 5),  #physical size, (width,height)
    title="Weekly TikTok Activity"
)

ax.set_xlabel('Week')
ax.set_ylabel('Videos Watched')

#remove time element in the weeks
ax.set_xticklabels([
    date.strftime("%Y-%m-%d")
    for date in weekly_counts.index
])

ax.plot()
```
